```python

#Inputs what number you are going to operate
num1 = float(input("Enter number 1: "))
num2 = float(input("Enter number 2: "))

#shows what operations you can choose
print("Choices for operations: ADDITION = 1, SUBTRACTION = 2, MULTIPLICATION = 3, DIVISION = 4")
operation = input("Choose an operation: ").strip().upper()

#This function Chooses an operation and executes it based on what operation you choose
def operation_chooser(operation, n1, n2):
    if operation == "1":
        print(f"Result: {n1 + n2}")
    elif operation == "2":
        print(f"Result: {n1 - n2}")
    elif operation == "3":
        print(f"Result: {n1 * n2}")
    elif operation == "4":
        if n2 != 0:
            print(f"Result: {n1 / n2}")
        else:
            print("Error: Division by zero is not allowed.")
    else:
        print("Invalid choice, please put in an actual choice.")


# Call the function and pass the arguments
operation_chooser(operation, num1, num2)

#I used my own functions and input to store and calculate the numbers
#The parameters i used were n1, n2 , operation
#whatever numbers were inputn and what operation they choose
#by storing them and when its time to operate it uses it
#Because it can save alot of time and make it look better
```
