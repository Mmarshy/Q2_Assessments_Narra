```python

num1 = float(input("Enter number 1: "))
num2 = float(input("Enter number 2: "))

choices = print("choices for operations is: ADDITION, SUBTRACTION , MULTIPLICATION, DIVISION")

operation = input("Choose an operation: ")

def operation_chooser(operation):
  if operation == ADDITION:
    def add_num(num1, num2):
      return num1 + num2

  print(add_num(num1, num2))

  elif operation == SUBTRACTION:
    def subtract_num(num1, num2):
      return num1 - num2

  print(subtract_num(num1, num2):)

  elif operation == MULTIPLICATION:
    def multiply_num(num1, num2):
      return num1 * num2

  print(multiply_num(num1, num2):)

  elif operation == DIVISION:
    def divide_num(num1, num2):
      return num1/num2

  print(divide_num(num1, num2):)

  else:
    print("invalid please put in an actual choice")





```
