```python
def show_score(scores):
    print("\nCurrent Scores:")
    for i in range(len(scores)):
        print("Index", i, ":", scores[i])

def search_score(scores):
    target = int(input("Enter score to search for: "))
    if target in scores:
        print("Score found in the list!")
    else:
        print("Score not found.")

def find_average(scores):
    total = sum(scores)
    average = total / len(scores)
    print("Average Score:", average)

def update_score(scores):
    show_score(scores)
    index = int(input("Enter the index number to update: "))
    new_value = int(input("Enter the new score: "))

    scores[index] = new_value
    print("Score updated successfully!")

def main():
    scores = [80, 90, 75, 85, 95]

    while True:
        print("\n=== Student Score Analyzer ===")
        print("1. Show Scores")
        print("2. Search Score")
        print("3. find Average Score")
        print("4. Update Score")
        print("5. Show Updated Scores")
        print("6. Exit")

        choice = input("Enter your choice: ")

        if choice == '1':
            show_score(scores)
        elif choice == '2':
            search_score(scores)
        elif choice == '3':
            find_average(scores)
        elif choice == '4':
            update_score(scores)
        elif choice == '5':
            show_score(scores)
        elif choice == '6':
            print("Goodbye!")
            break
        else:
            print("Invalid choice. Try again.")

main()
```
